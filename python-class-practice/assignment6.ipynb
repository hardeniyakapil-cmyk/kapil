```python
#6.6. Write a program to separate odd and even keys from a dictionary.
# Also count the total number of odd keys and even keys.
d = {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}

odd_keys = {}
even_keys = {}

odd_count = 0
even_count = 0

for key, value in d.items():
    if key % 2 == 0:
        even_keys[key] = value
        even_count += 1
    else:
        odd_keys[key] = value
        odd_count += 1

print("Odd Keys Dictionary:", odd_keys)
print("Even Keys Dictionary:", even_keys)
print("Total Odd Keys:", odd_count)
print("Total Even Keys:", even_count)
```

```python
# 16. Write a program to count the frequency of each character in a string using a
# dictionary.
# Example: "banana"
s = "banana"

freq = {}

for ch in s:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print("Character Frequency:", freq)
```

```python
# 24. Write a program to remove duplicate values from a dictionary.
d = {"A": 10, "B": 20, "C": 10, "D": 30, "E": 20}

new_dict = {}

for key, value in d.items():
    if value not in new_dict.values():
        new_dict[key] = value

print(new_dict)
```

```python
# 25. Write a program to check whether all values in a dictionary are unique
d = {"A": 10, "B": 20, "C": 30, "D": 20}

if len(d.values()) == len(set(d.values())):
    print("All values are unique.")
else:
    print("Duplicate values found.")
```
